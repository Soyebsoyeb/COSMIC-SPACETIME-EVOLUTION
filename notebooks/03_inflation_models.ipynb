# Inflation models

Exact e-fold background evolution for two potentials. Label: **model dependent**.

```python
import numpy as np
import matplotlib.pyplot as plt
from cosmic_evolution.early_universe import inflation as inf

for model in (inf.StarobinskyModel(), inf.QuadraticModel()):
    sol = inf.solve_background(model)
    n = 55.0
    phi = sol.phi_at(n)
    sr = inf.slow_roll_observables(model, phi)
    hs = inf.hubble_slow_roll_observables(sol, n)
    print(model.name, 'N_total = %.1f' % sol.N_total, 'n_s = %.4f' % hs['n_s'], 'r = %.4f' % hs['r'], '| potential slow roll:', round(sr['n_s'], 4), round(sr['r'], 4))
```

```python
sol = inf.solve_background(inf.StarobinskyModel())
plt.semilogy(sol.N_before_end(), sol.eps_H); plt.gca().invert_xaxis()
plt.xlabel('e-folds before end'); plt.ylabel('epsilon_H'); plt.show()
```

The quadratic potential gives $r\approx0.14$, above the observational upper limit of order $0.06$: the code reproduces this known tension, it does not hide it.

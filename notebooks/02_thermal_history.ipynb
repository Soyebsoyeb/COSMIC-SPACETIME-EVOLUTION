# Thermal history

Relativistic degrees of freedom and the time-temperature relation. Label: **approximate** (step function at T = m, instantaneous decoupling).

```python
import numpy as np
import matplotlib.pyplot as plt
from cosmic_evolution.config import CosmologyParams
from cosmic_evolution.early_universe import thermal_history as th

T = np.logspace(-6, 3, 400)
plt.semilogx(T, th.g_star(T), label='g*')
plt.semilogx(T, th.g_star_s(T), '--', label='g*s')
plt.xlabel('T [GeV]'); plt.ylabel('degrees of freedom'); plt.legend(); plt.show()
```

```python
for name, Tg, t in th.epoch_table(CosmologyParams()):
    print(f'{name:50s} T = {Tg:9.3g} GeV   t = {t:9.3g} s')
```
